```python
import random
game =['Rock', 'Paper', 'Scissors']
you=input("enter your choice").capitalize()
p1=random.choice(game).capitalize()
y=0
p=0
n=int(input("How many rounds of match: "))
for i in range(n):
  if(you==p1):
    print("Tie")
  elif((you=='Rock' and p1=='Scissors')or (you=='Scissors' and p1=='Paper')or (you=='Paper' and p1=='Rock')):
    print("Point to You")
    y+=1
  elif((p1=='Rock' and you=='Scissors')or (p1=='Scissors' and you=='Paper')or (p1=='Paper' and you=='Rock')):
    print("Point to p1")
    p1+=1

if(p>y):
  print(f'YOu ARE THE WINNER WITH POINTS {y}')

elif(p<y):
  print(f'P1 IS THE WINNER WITH POINTS {p}')
```

```python
class rectangle: #class
  def __init__(self,length,width):  #constructor
    self.length = length
    self.width = width

  def Area(self):  #object : replica to call the methods of the class or to implement or access it
    area = self.length * self.width
    return area

  def perimeter(self):
    per = 2*(self.length + self.width)
    return per

rec = rectangle(2,4)
print(rec.Area())
print(rec.perimeter())
```

```python
# number guessing game
import random
ch=0
def user_choice():
  ch=int(input("Enter your choice:[1 OR 2 OR 3]"))
  if(ch==1):
    print("You're at easy level\nATTEMPTS : 5")
  elif(ch==2):
    print("You're at medium level\nATTEMPTS : 7")
  elif(ch==3):
    print("You're at hard level\nATTEMPTS : 10")
  else:
    print("can't you see the instructions")
  return ch

def num_guess():
  ch=user_choice()
  if(ch==1):
    max_attempt=5
    max_num = 10
  elif(ch==2):
    max_attempt=7
    max_num=50
  elif(ch==3):
    max_attempt=10
    max_num=100
  else:
    print('Wrong choice exiting the game')
    exit()
  for i in range(max_attempt):
    num=random.randint(1,11)
    guess=int(input("Guess the number :"))
    print("Random number is: ",num)
    max_attempt=max_attempt-1
    if(num==guess):
      print("You GUESSED THE NUMBER CORRECT")
      break
    elif(num>guess):
      print('TOO LOW! Try Again Mate\n''attempts remaining',max_attempt)
    elif(num<guess):
      print('TOO HIGH! Try Again Mate\n''attempts remaining',max_attempt)
    else:
      print('attempts remaining',max_attempt)
    if(max_attempt==0):
      print('Game Over! The correct number was:',num)
while(1):
  print("number guess game")
  num_guess()
  choice=input("Do you want to try again?\nenter your choice: [Y/N] ")
  if(choice in ('Y', 'y')):
    num_guess()
  else:
    break
```
